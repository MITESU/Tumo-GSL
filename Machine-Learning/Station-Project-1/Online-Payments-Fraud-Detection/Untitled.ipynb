```python
import pandas as pd
import numpy as np
import plotly.express as px
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

read_df = pd.read_csv("PS_20174392719_1491204439457_log.csv")
#read_df.head()
#read_df.info()

read_df = read_df.drop(columns=["nameOrig", "nameDest"])
read_df = pd.get_dummies(read_df, columns=["type"]) #text to numbers
fig_scatter = px.scatter(
    read_df,
    x="amount",
    y="oldbalanceOrg",
    color="isFraud",
    title="Amount vs OldBalance"
)
fraud_counts = read_df["isFraud"].value_counts()

fig_pie = px.pie(
    values=fraud_counts.values,
    names=["Not Fraud", "Fraud"],
    title="Fraud Distribution"
)

fig_scatter.show()
fig_pie.show()

X = read_df.drop(columns=["isFraud"])
y = read_df["isFraud"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = LinearRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_test)
predictions_binary = [1 if p >= 0.5 else 0 for p in predictions] #to binary

score = accuracy_score(y_test, predictions_binary)

print("Predictions and Score accuracy: ", predictions_binary[:10], score) #predictions_bool[:10] -> first 10 predictions mda
```
